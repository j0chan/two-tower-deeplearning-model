```python
import numpy as np
import faiss
import os
```

```python
save_dir = './saved_models/final_model'
embeddings_path = os.path.join(save_dir, 'all_item_embeddings.npy')
index_path = os.path.join(save_dir, 'faiss_index.bin')
```

```python
# 사전 계산된 가게 벡터 로드
print(f"'{embeddings_path}' 에서 가게 벡터를 로드")
all_item_embeddings = np.load(embeddings_path)
print('완료')
```

```python
# 벡터의 차원(dimension) 확인
d = all_item_embeddings.shape[1]
print(f"가게 벡터의 차원: {d}")
```

```python
# FAISS 인덱스 생성
# IndexFlatIP: 내적(Inner Product)을 기준으로 거리를 계산하는 인덱스
index = faiss.IndexFlatIP(d)
print(f"FAISS 인덱스(IndexFlatIP) 생성 완료.")
```

```python
# 인덱스에 벡터 추가
index.add(all_item_embeddings)
print(f"{index.ntotal}개의 가게 벡터를 인덱스에 추가 완료.")
```

```python
# 생성된 인덱스 저장
faiss.write_index(index, index_path)
print(f"생성된 FAISS 인덱스를 '{index_path}' 경로에 저장 완료.")
```
